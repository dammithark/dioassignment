```python
import gradio as gr

KB = {
    ('hello', 'hi', 'namaste'): "Namaste! Ask me about AI, machine learning, image processing or prompt engineering.",
    ('machine learning', ' ml'): "Machine learning lets systems learn patterns from data instead of following hand-written rules.",
    ('deep learning', 'neural'): "Deep learning uses multi-layer neural networks to learn features automatically from raw data.",
    ('blend', 'opencv', 'image'): "Image blending combines two images as alpha*img1 + beta*img2 + gamma, e.g. with cv2.addWeighted.",
    ('prompt',): "Good prompts specify subject, style, camera/format, constraints and what to avoid (negative prompts).",
    ('pandas', 'dataframe'): "Pandas provides DataFrames for loading, cleaning, grouping and exporting tabular data.",
    ('bye', 'thank'): "You're welcome! Good luck with the GOT-A assessment.",
}

def respond(message, history):
    m = ' ' + message.lower()
    for keys, answer in KB.items():
        if any(k in m for k in keys):
            return answer
    return ("I'm a simple demo assistant. Try asking about machine learning, deep learning, "
            "image blending, prompt engineering or pandas.")

demo = gr.ChatInterface(
    fn=respond,
    title="GOT-A AI Learning Assistant",
    description="A demo chatbot for the NIELIT GOT-A program. Type a question or click a sample prompt below.",
    examples=["What is machine learning?",
              "How does image blending work?",
              "Give me tips for prompt engineering"],
)
demo.launch()   # open http://127.0.0.1:7860 in a browser and take the screenshot there
```
