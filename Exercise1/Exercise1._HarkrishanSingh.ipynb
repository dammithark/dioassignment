```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. DataFrame of random values (seeded so results are reproducible)
rng = np.random.default_rng(42)
df = pd.DataFrame(rng.uniform(0, 100, size=(100, 4)).round(2),
                  columns=['f1', 'f2', 'f3', 'f4'])
df.head()
```

```python
# 3. Row-wise mean
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1).round(2)

# 4. Export to Excel (needs openpyxl) and verify by reading back
df.to_excel('Exercise1_output.xlsx', index=False)
pd.read_excel('Exercise1_output.xlsx').head()
```
